# Part 3:  Create Coordinates for A and B

Use the three original lists to create two coordinate lists called `point_A` and `point_B`.  Each list should contain the Easting and Northing of the corresponding location, in that order:

```python
point_A = [easting, northing]
point_B = [easting, northing]
```

Create these without hardcoding (manually typing in) the coordinate numbers.

Instead, retrieve the coordinates from the `eastings` and `northings` lists, using the appropriate indexes.

**Print** both coordinate lists.

````python
# Your answer to Part 3 using code:
def reformat_coords(locations, search):
    index = 0
    point = []
    for location in locations:
        if location == search:
            easting = eastings[index]
            northing = northings[index]
            point.append(int(easting))
            point.append(int(northing))
            print(f"point_{search} = {point}")
            return point
        index += 1

reformat_coords(locations, "A")
reformat_coords(locations, "B")
````

# Part 4: Calculate the Distance from A to B

Create a function named `calculate_distance` that has two parameters:

```python
calculate_distance(point1, point2)
```

The function should calculate the straight-line distance between two projected coordinates:

distance(A,B) $= \sqrt{(A_\text{easting} - B_\text{easting})^2 +  (A_\text{northing} - B_\text{northing})^2}$

You may want to use the python exponent operator `**`, and to remember that the square root of a number is the number to the power $\frac{1}{2}:$ 


```python
A_squared = A**2 #this is the square of A
A_sqrt = A**0.5 #this is the square root of A  
```

The function should **return** the calculated distance.

Use the function to calculate the distance between A and B, and *print* the result.

````python
# Your answer to Part #4 using code: 
# calculating the distance from A to B requires separate steps and the function needs to have two parameters
## step one: subtract A easting and B easting
## step two: square the result of that
## step 3: subtract the A and B northings
## step four: square the result of that
## step five: add those results together
## step six: take the square root of that entire thing

def calculate_distance(point1, point2):
    x = 0
    distance = ((point1[x] - point2[x])**2) + ((point1[x + 1] - point2[x + 1])**2) ** .5
    return distance

calculate_distance(reformat_coords(locations, "A"), reformat_coords(locations, "B"))
````
