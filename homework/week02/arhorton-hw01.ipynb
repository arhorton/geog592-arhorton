# Part 3:  Create Coordinates for A and B

# Part 7 — Distances from A to B, C, and D

Now use the original three lists to calculate all of the following distances between locations: A to B, A to C, and A to D.

Do not manually type the coordinates for each location.  Use the original `locations`, `eastings`, and `northings` lists, and your `calculate_distance()` function.

Your output should clearly identify which distance is being printed, for example:

```text
Distance from A to D: ...
```

````python
# Your answer to Part #7 using code: 
## creating the points that I need to use for later
def reformat_coords(locations, search):
    index = 0
    point = []
    for location in locations:
        if location == search:
            easting = eastings[index]
            northing = northings[index]
            point.append(int(easting))
            point.append(int(northing))
            print(f"point_{search} = {point}")
            return point
        index += 1

a = reformat_coords(locations, "A")
b = reformat_coords(locations, "B")
c = reformat_coords(locations, "C")
d = reformat_coords(locations, "D")


# now calculating the three distances
def calculate_distance(point1, point2):
    locational_points = [a, b, c, d]
    try: 
        if not isinstance(point1, list) or not isinstance(point2, list):
            raise ValueError
        if len(point1) != 2 or len(point2) != 2:
            raise IndexError
        if not all(isinstance(x, (int, float)) for x in point1):
            raise TypeError
        if not all(isinstance(x, (int, float)) for x in point2):
            raise TypeError
        x = 0
        distance = 0
        while x < len(point1):
            location_1 = locations[locational_points.index(point2)]
            distance += ((point1[x] - point2[x])**2)
            x += 1
        distance = distance ** .5
        print(f"Distance from A to {location_1}: {distance}")
    except IndexError:
        return "List length is incorrect!"
    except TypeError:
        return "The lists are in the wrong format!"
    except ValueError:
        return "Inputs are not lists!"


calculate_distance(a, b)
calculate_distance(a, c)
calculate_distance(a, d)
# the last printed output is the error that the function throws when the conditions become untrue
````
