### The first task is to import a CSV!
I chose brownfields in North Carolina as my CSV. It has a little over 1,000 records and depicts all of the sites in NC where CERCLA dictates that there are hazardous levels of contaminants in the soil/water/environment. 

```python
## importing the CSV

import csv

filename = "brownfields.csv"

with open(filename, "r") as file:
    reader = csv.DictReader(file)
    rows = list(reader)

print("Total number of records: ", len(rows))
print("This is going to be great!")
```

### The second task is to clean said CSV!
I am going to filter out any values that are None or missing from any of the columns; x and y will be evaluated to see if the values are numeric (it will be helpful for future connections to geographic data). Extra columns, meaning columns that contain information that I won't be readily working with, will be removed to ensure that the data is the easiest to work with it can be. The final, cleaned CSV will be exported and stored in the homework/week04/data folder as instructed. 

```python
## removing all of the unneccessary columnns that I don't need
cols_needed = ["BF_Name", "BF_Acreage", "x2", "y2"]
extracted_bf = [{col: row[col] for col in cols_needed} for row in rows]

for row in extracted_bf: # this loop allows me to see the output of the smaller dataset that I have just created
    print(row)


## I've now created a dataset with just the names, acerage, and geographic coordinates of each brownfield in NC
```

```python
## checking for the rows that are None
None in extracted_bf # I found this code on stack overflow that evaluates to a boolean that declares if any values are None

## checking for the rows that are missing a value
for i, row in enumerate(extracted_bf):
    empty_vals = []
    if row.get("BF_Name") == "" or row.get("BF_Acreage") == "" or row.get("x2") == "" or row.get("y2") == "": 
        empty_vals.append(i)

extracted_filtered = [i for i in extracted_bf if i not in empty_vals]

not_num = []
for row in extracted_filtered: ## the correctly filtered and cleaned dataset
    try: 
        float(row["x2"]) # making sure that both the x and y coordinates of the values are numeric
        float(row["y2"])
    except ValueError:
        not_num.append(row["x2"])
        not_num.append(row["y2"])

print(not_num) # if correct, this should be an empty list
print(extracted_filtered)
print(len(extracted_filtered))
```

### The third task is to export the cleaned CSV!

```python
fieldnames = extracted_filtered[0].keys()

with open("cleaned_brownfields.csv", "w", newline="") as file: 
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(extracted_filtered)
```
